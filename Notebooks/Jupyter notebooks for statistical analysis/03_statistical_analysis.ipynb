# 03 Statistical Analysis

This notebook runs statistical models such as ANOVA and regression to evaluate fairness effects.

```
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/Fairness_Raw_Data.csv")
df = pd.read_csv(DATA_PATH)
df.head()
```

```
import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.ols("Overall_Score ~ C(Model) + C(Prompt) + C(L1) + C(CEFR)", data=df).fit()
print(model.summary())
```
