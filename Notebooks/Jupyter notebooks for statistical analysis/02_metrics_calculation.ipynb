# 02 Metrics Calculation

This notebook demonstrates how FQI and MFI can be calculated from the model feedback scores.

```
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/Fairness_Raw_Data.csv")
df = pd.read_csv(DATA_PATH)
df.head()
```

```
# Example aggregation by Model and Prompt
score_cols = ["Accuracy", "Helpfulness", "Specificity", "Actionability", "Pedagogical_Appropriateness", "Socio_Affective_Tone", "Overall_Score"]
summary = df.groupby(["Model", "Prompt"])[score_cols].mean().reset_index()
summary
```
