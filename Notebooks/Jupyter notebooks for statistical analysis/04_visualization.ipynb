# 04 Visualization

This notebook creates figures for the manuscript and saves them in the figures directory.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/Fairness_Raw_Data.csv")
df = pd.read_csv(DATA_PATH)
df.head()
```

```
fig, ax = plt.subplots(figsize=(8, 4))
df.groupby("Prompt")["Overall_Score"].mean().plot(kind="bar", ax=ax)
ax.set_title("Overall Score by Prompt")
ax.set_ylabel("Mean Overall Score")
plt.tight_layout()
plt.show()
```
